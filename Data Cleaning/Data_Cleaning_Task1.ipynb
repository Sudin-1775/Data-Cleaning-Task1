### Data Cleaning and Preprocessing

#### 1. Handle Duplicate Rows

```python
print(f"Number of rows before removing duplicates: {df_excel.shape[0]}")
df_excel.drop_duplicates(inplace=True)
print(f"Number of rows after removing duplicates: {df_excel.shape[0]}")
```

#### 2. Clean Categorical Columns

```python
categorical_cols = ["Product", "Category", "Region", "City", "Payment_Method", "Sales_Channel", "Customer_Type"]

for col in categorical_cols:
    if col in df_excel.columns and df_excel[col].dtype == 'object':
        df_excel[col] = df_excel[col].astype(str).str.strip().str.lower().replace({'nan': None})
        print(f"Cleaned unique values for {col}:\n{df_excel[col].dropna().unique()}\n")
```

#### 3. Convert 'Discount' to Numeric

```python
print("Original Discount Dtype:", df_excel['Discount'].dtype)
df_excel['Discount'] = pd.to_numeric(df_excel['Discount'], errors='coerce')
print("New Discount Dtype:", df_excel['Discount'].dtype)
print("Number of NaN values in Discount after conversion:", df_excel['Discount'].isnull().sum())
```

#### 4. Handle Missing Values

```python

print("Missing values before imputation:")
print(df_excel.isnull().sum())

df_excel['Order_Date'] = df_excel['Order_Date'].ffill()

df_excel['Quantity'] = df_excel['Quantity'].fillna(df_excel['Quantity'].median())

df_excel['Discount'] = df_excel['Discount'].fillna(df_excel['Discount'].median())

for col in ['City', 'Payment_Method']:
    if df_excel[col].isnull().any():
        mode_val = df_excel[col].mode()[0]
        df_excel[col] = df_excel[col].fillna(mode_val)

print("\nMissing values after imputation:")
print(df_excel.isnull().sum())
```

```python

# TASK 1 - DATA CLEANING

cleaned_df = df_excel.copy()

print("Starting rows:", len(cleaned_df))

# STEP 1: Remove duplicate records

duplicates_removed = cleaned_df.duplicated().sum()

cleaned_df = cleaned_df.drop_duplicates()

print("Duplicate rows removed:", duplicates_removed)

# STEP 2: Clean text columns

text_columns = [
    "Product",
    "Category",
    "Region",
    "City",
    "Payment_Method",
    "Sales_Channel",
    "Customer_Type"
]

for column in text_columns:
    cleaned_df[column] = (
        cleaned_df[column]
        .astype("string")
        .str.strip()
    )


# STEP 3: Standardize Product names

product_mapping = {
    "laptop": "Laptop",
    "Laptop": "Laptop",
    "smart phone": "Smartphone",
    "Smartphone": "Smartphone",
    "tablet": "Tablet",
    "monitor": "Monitor",
    "keyboard": "Keyboard",
    "mouse": "Mouse",
    "headphones": "Headphones",
    "printer": "Printer",
    "office chair": "Office Chair",
    "desk": "Desk",
    "webcam": "Webcam",
    "external ssd": "External SSD"
}

cleaned_df["Product"] = (
    cleaned_df["Product"]
    .str.lower()
    .map(product_mapping)
    .fillna(cleaned_df["Product"].str.title())
)

# STEP 4: Standardize Category

category_mapping = {
    "electronics": "Electronics",
    "accessories": "Accessories",
    "furniture": "Furniture",
    "office equipment": "Office Equipment",
    "storage": "Storage"
}

cleaned_df["Category"] = (
    cleaned_df["Category"]
    .str.lower()
    .map(category_mapping)
    .fillna(cleaned_df["Category"].str.title())
)


# STEP 5: Standardize Region
cleaned_df["Region"] = (
    cleaned_df["Region"]
    .str.strip()
    .str.title()
)

# STEP 6: Standardize City

cleaned_df["City"] = (
    cleaned_df["City"]
    .str.strip()
    .str.title()
)


# STEP 7: Standardize Payment Method

cleaned_df["Payment_Method"] = (
    cleaned_df["Payment_Method"]
    .str.strip()
    .str.title()
)

# STEP 8: Standardize Sales Channel

cleaned_df["Sales_Channel"] = (
    cleaned_df["Sales_Channel"]
    .str.strip()
    .str.title()
)

# STEP 9: Standardize Customer Type

cleaned_df["Customer_Type"] = (
    cleaned_df["Customer_Type"]
    .str.strip()
    .str.title()
)

# STEP 10: Clean Order Date

cleaned_df["Order_Date"] = pd.to_datetime(
    cleaned_df["Order_Date"],
    errors="coerce",
    format="mixed"
)

# STEP 11: Clean Quantity

cleaned_df["Quantity"] = pd.to_numeric(
    cleaned_df["Quantity"],
    errors="coerce"
)

quantity_median = cleaned_df["Quantity"].median()

cleaned_df["Quantity"] = cleaned_df["Quantity"].fillna(
    quantity_median
)

cleaned_df["Quantity"] = cleaned_df["Quantity"].round().astype(int)


# ------------------------------------------
# STEP 12: Clean Unit Price
# ------------------------------------------

cleaned_df["Unit_Price"] = pd.to_numeric(
    cleaned_df["Unit_Price"],
    errors="coerce"
)


# STEP 13: Clean Discount

cleaned_df["Discount"] = (
    cleaned_df["Discount"]
    .astype("string")
    .str.replace("%", "", regex=False)
)

cleaned_df["Discount"] = pd.to_numeric(
    cleaned_df["Discount"],
    errors="coerce"
)

cleaned_df.loc[
    cleaned_df["Discount"] > 1,
    "Discount"
] = cleaned_df.loc[
    cleaned_df["Discount"] > 1,
    "Discount"
] / 100

cleaned_df["Discount"] = cleaned_df["Discount"].fillna(0)

# STEP 14: Clean Sales

cleaned_df["Sales"] = pd.to_numeric(
    cleaned_df["Sales"],
    errors="coerce"
)

# STEP 15: Handle missing City and Payment

cleaned_df["City"] = cleaned_df["City"].fillna("Unknown")
cleaned_df["City"] = cleaned_df["City"].replace("", "Unknown")

cleaned_df["Payment_Method"] = cleaned_df["Payment_Method"].fillna("Unknown")
cleaned_df["Payment_Method"] = cleaned_df["Payment_Method"].replace("", "Unknown")

# STEP 16: Remove rows with missing dates

rows_before_date_cleaning = len(cleaned_df)

cleaned_df = cleaned_df.dropna(
    subset=["Order_Date"]
)

print(
    "Rows removed because of invalid/missing dates:",
    rows_before_date_cleaning - len(cleaned_df)
)

# STEP 17: Recalculate Sales

cleaned_df["Sales"] = (
    cleaned_df["Quantity"]
    * cleaned_df["Unit_Price"]
    * (1 - cleaned_df["Discount"])
).round(2)

# STEP 18: Sort by Order Date

cleaned_df = cleaned_df.sort_values(
    "Order_Date"
).reset_index(drop=True)


print("\n========== CLEANING COMPLETED ==========")
print("Final rows:", len(cleaned_df))
print("Final columns:", len(cleaned_df.columns))

print("\nMissing values after cleaning:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows after cleaning:")
print(cleaned_df.duplicated().sum())

print("\nCleaned dataset preview:")
display(cleaned_df.head(10))
```
